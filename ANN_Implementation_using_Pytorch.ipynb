```python
import torch
from torch import nn

class ANN(nn.Module):
    def __init__(self, input_size, hidden_size, output_size):
        super(ANN, self).__init__()
        # First fully connected layer
        self.fc1 = nn.Linear(input_size, hidden_size)
        # Activation function
        self.relu = nn.ReLU()
        # Output layer
        self.fc2 = nn.Linear(hidden_size, output_size)

    def forward(self, x):
        # Flatten input (like image)
        x = x.view(x.size(0), -1)
        x = self.fc1(x)
        x = self.relu(x)
        x = self.fc2(x)
        return x

# Initialize model
model = ANN(input_size=784, hidden_size=128, output_size=10)
```

###Loss Function and Optimizer

```python
import torch.optim as optim
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=0.001)
```

###Training Loop

```python
epochs = 10
for epoch in range(epochs):
    model.train()
    batch_size = 3
    inputs = torch.rand(3,28,28)
    target = torch.randint(0, 9, (batch_size, 9))
    # Forward pass: compute predicted outputs
    outputs = model(inputs)
    #loss = criterion(outputs, target.view(-1))
    loss = criterion(outputs, torch.rand(3))
    # Backward pass: compute gradients
    optimizer.zero_grad() # Clear previous gradients
    loss.backward()

    # Optimization step: update weights
    optimizer.step()

    print(f"Epoch [{epoch+1}/{epochs}], Loss: {loss.item():.4f}")
```
